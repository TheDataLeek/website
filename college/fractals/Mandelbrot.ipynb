```python
import numpy as np
from calculate_iterations import calculate_iterations
import matplotlib.pyplot as plt
from tqdm import tqdm

def build_image(name="Mandelbrot.png", max_iterations=50, width=1000):
    height = int(width * (2/3.5))
    
    colors = np.zeros((height, width))
    
    def add_color():
        x0s = [((3.5 * px)/width) - 2.5 for px in range(width)]
        y0s = [((2 * py)/height) - 1 for py in range(height)]
        for px in tqdm(range(width)):
            x0 = x0s[px]
            for py in range(height):
                y0 = y0s[py]
                colors[py][px] = (255 * calculate_iterations(x0, y0, max_iterations))/max_iterations
                
    add_color()
    plt.imshow(colors)
    plt.imsave(name, colors)
```

```python
build_image()
```

```python
build_image(name="MandelbrotBig.png", max_iterations=50, width=5_000)
```
